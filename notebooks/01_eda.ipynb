# Retail Sales Analysis — EDA Notebook

Exploratory analysis of the retail sales dataset.

**Workflow:**
1. Load & validate the cleaned dataset
2. Univariate analysis (distributions)
3. Time-series trends
4. Category / region / channel breakdowns
5. Return-rate & discount analysis
6. Key takeaways

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

plt.rcParams.update({'figure.dpi': 120})

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
df = pd.read_csv(ROOT / 'data' / 'processed' / 'retail_sales_clean.csv', parse_dates=['order_date'])
print(f'{len(df):,} rows x {df.shape[1]} columns')
df.head()
```

## 1. Data validation

```python
df.info()
df.describe(include='all').T
```

```python
df.isna().sum()[df.isna().sum() > 0]
```

## 2. Distributions

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 3.5))
df['net_revenue'].plot(kind='hist', bins=50, ax=axes[0], title='Net Revenue per Order')
df['quantity'].plot(kind='hist', bins=10, ax=axes[1], title='Units per Order')
plt.tight_layout()
```

## 3. Time-series trend

```python
monthly = df.set_index('order_date')['net_revenue'].resample('ME').sum()
monthly.plot(figsize=(10, 3.5), title='Monthly Net Revenue', ylabel='$')
plt.tight_layout()
```

## 4. Breakdowns

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 3.5))
df.groupby('category')['net_revenue'].sum().sort_values().plot(kind='barh', ax=axes[0], title='Revenue by Category')
df.groupby('region')['net_revenue'].sum().sort_values().plot(kind='barh', ax=axes[1], title='Revenue by Region')
df.groupby('channel')['net_revenue'].sum().sort_values().plot(kind='barh', ax=axes[2], title='Revenue by Channel')
plt.tight_layout()
```

## 5. Returns & discounts

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 3.5))
df.groupby('category')['returned'].mean().mul(100).sort_values(ascending=False).plot(
    kind='bar', ax=axes[0], rot=30, title='Return Rate by Category (%)', ylabel='%')
df.groupby('discount_pct')['net_revenue'].mean().plot(
    kind='bar', ax=axes[1], rot=0, title='Avg Net Revenue by Discount %')
plt.tight_layout()
```

## 6. Key takeaways

- Electronics dominates revenue (~50% of total) with the highest AOV.
- Online and In-Store channels contribute comparable revenue — check regional mix for channel strategy.
- Return rates are stable at ~8%; monitor Beauty and Home & Kitchen.
- Deeper discounts do not clearly lift average order value in this data.
